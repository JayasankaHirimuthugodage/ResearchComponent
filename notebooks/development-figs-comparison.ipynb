# FIGS development-only comparison

This notebook reads the completed aggregate FIGS report only. It does not fit models or load participant-level or final-test data.

```python
from pathlib import Path
import json
import pandas as pd
root = Path.cwd(); root = root.parent if root.name == 'notebooks' else root
report = json.loads((root / 'reports/generated/development-figs-comparison-report.json').read_text())
assert report['test_cohort_loaded'] is False and report['final_test_evaluated'] is False
assert report['development_participants'] == 500 and report['feature_count'] == 16
pd.Series({'status': report['status'], 'runtime seconds': report['runtime_seconds'], 'outer folds/domain': report['cross_validation']['outer_evaluations_per_domain'], 'imodels version': report['imodels_version']})
```

```python
rows = []
for domain, value in report['domains'].items():
    m = value['repeat_metric_summary']
    rows.append({'Domain': domain, 'PR-AUC': m['average_precision_pr_auc']['mean'], 'AUROC': m['roc_auc']['mean'], 'Accuracy': m['accuracy']['mean'], 'Balanced accuracy': m['balanced_accuracy']['mean'], 'Sensitivity': m['sensitivity']['mean'], 'Specificity': m['specificity']['mean'], 'Precision': m['precision']['mean'], 'NPV': m['negative_predictive_value']['mean'], 'F1': m['f1']['mean'], 'Brier': m['brier_score']['mean']})
pd.DataFrame(rows).set_index('Domain').round(4)
```

```python
pd.DataFrame([{'Domain': d, 'Mean rule representations': v['outer_model_complexity']['rule_count']['mean'], 'Rule range': f"{v['outer_model_complexity']['rule_count']['minimum']}–{v['outer_model_complexity']['rule_count']['maximum']}", 'Mean trees': v['outer_model_complexity']['tree_count']['mean']} for d, v in report['domains'].items()])
```

## Interpretation

Values are means of five complete repeat-level out-of-fold evaluations. Historical and TabICLv2 comparisons are numerical only because aligned predictions are unavailable for paired inference. FIGS remains a development-only interpretable comparator; this report is not independent validation, clinical diagnostic evidence, or a deployment decision.
