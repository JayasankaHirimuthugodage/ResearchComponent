# LightGBM development-only comparison

Aggregate-only view of the fixed LightGBM experiment. This notebook loads only the completed aggregate report; it does not fit a model or load participant or final-test data.

```python
from pathlib import Path
import json
import pandas as pd
root = Path.cwd(); root = root.parent if root.name == 'notebooks' else root
report = json.loads((root / 'reports/generated/development-lightgbm-comparison-report.json').read_text())
assert report['test_cohort_loaded'] is False
assert report['development_participants'] == 500 and report['feature_count'] == 16
pd.Series({'status': report['status'], 'LightGBM version': report['lightgbm_version'], 'runtime seconds': report['runtime_seconds'], 'outer folds/domain': report['cross_validation']['outer_evaluations_per_domain']})
```

```python
rows = []
for domain, value in report['domains'].items():
    metrics = value['repeat_metric_summary']
    rows.append({'Domain': domain, 'PR-AUC': metrics['average_precision_pr_auc']['mean'], 'AUROC': metrics['roc_auc']['mean'], 'Accuracy': metrics['accuracy']['mean'], 'Balanced accuracy': metrics['balanced_accuracy']['mean'], 'Sensitivity': metrics['sensitivity']['mean'], 'Specificity': metrics['specificity']['mean'], 'Brier': metrics['brier_score']['mean']})
pd.DataFrame(rows).set_index('Domain').round(4)
```

## Interpretation

The configured LightGBM candidate is an exploratory gradient-boosted-tree comparator. Metrics are means of five repeat-level out-of-fold estimates. Historical comparisons are numerical only because aligned participant-level predictions were not retained. This result does not establish superiority, deployment readiness, or performance on the locked 125-participant test cohort.
