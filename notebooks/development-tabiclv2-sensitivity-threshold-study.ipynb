# TabICLv2 sensitivity-constrained threshold study

Aggregate-only development report viewer. The study is post-hoc exploratory, uses no test cohort, and retains no participant-level predictions.

```python
import json
from pathlib import Path

report = json.loads((Path.cwd().parent / 'reports/generated/development-tabiclv2-sensitivity-threshold-report.json').read_text())
assert report['cross_validation']['outer_evaluations_completed'] == 50
assert report['test_cohort_loaded'] is False
for domain, result in report['domains'].items():
    print(domain)
    for strategy, values in result['methods'].items():
        metric = values['participant_averaged_oof_metrics']
        print(strategy, {key: round(metric[key], 4) for key in ('sensitivity', 'specificity', 'accuracy', 'balanced_accuracy', 'average_precision_pr_auc')})
```
