# HistGradientBoostingClassifier development-only comparison

Aggregate-only rendering of one exploratory gradient-boosted-tree comparison. No participant records, model fitting, or final-test data are loaded.

```python
from pathlib import Path
import json, pandas as pd
root=Path.cwd(); root=root.parent if root.name=='notebooks' else root
report=json.loads((root/'reports/generated/development-histgradientboosting-comparison-report.json').read_text())
assert report['test_cohort_loaded'] is False and report['development_participants']==500
pd.Series({'status':report['status'],'runtime seconds':report['runtime_seconds'],'outer folds/domain':report['cross_validation']['outer_evaluations_per_domain']})
```

```python
rows=[]
for domain,value in report['domains'].items():
 m=value['repeat_metric_summary']; rows.append({'Domain':domain,'PR-AUC':m['average_precision_pr_auc']['mean'],'AUROC':m['roc_auc']['mean'],'Accuracy':m['accuracy']['mean'],'Balanced accuracy':m['balanced_accuracy']['mean'],'Sensitivity':m['sensitivity']['mean'],'Specificity':m['specificity']['mean'],'Brier':m['brier_score']['mean']})
pd.DataFrame(rows).set_index('Domain').round(4)
```

```python
rankings=[]
for domain,value in report['domains'].items():
 for row in value['eight_model_pr_auc_ranking']:
  rankings.append({'Domain':domain,'Candidate':row['candidate'],'Mean PR-AUC':row['mean_average_precision'],'Source':row['source']})
pd.DataFrame(rankings).sort_values(['Domain','Mean PR-AUC'],ascending=[True,False]).groupby('Domain').head(9).round(4)
```

## Interpretation

HistGradientBoostingClassifier is a gradient-boosted decision-tree comparator, not a wholly new algorithm family. Values are repeated outer-validation means; comparisons with historical models are numerical because aligned participant-level predictions for paired inference were not retained. This development-only result does not establish superiority, independent validation, clinical diagnostic accuracy, or a deployment decision.
